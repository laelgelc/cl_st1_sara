## Required Python packages

- pandas

## Importing the required libraries

````python
import pandas as pd
import os
import sys
````

## Define input variables

````python
input_directory = 'cl_st1_ph2_sara'
output_directory = 'cl_st1_ph3_sara'
id = 'short_stories'
dataset_filename_1 = f"{id}_sample"
dataset_filename_2 = f"{id}_corpus"
````

## Create output directory

````python
# Check if the output directory already exists. If it does, do nothing. If it doesn't exist, create it.
if os.path.exists(output_directory):
    print('Output directory already exists.')
else:
    try:
        os.makedirs(output_directory)
        print('Output directory successfully created.')
    except OSError as e:
        print('Failed to create the directory:', e)
        sys.exit(1)
````

## Import the data into a DataFrame

````python
df_short_stories_sample_balanced = pd.read_json(f"{input_directory}/{dataset_filename_1}.jsonl", lines=True)
````

````python
df_short_stories_sample_balanced
````

## Dropping unnecessary columns

````python
columns = df_short_stories_sample_balanced.columns.tolist()
print(columns)
````

````python
needed_columns = [
    'url',
    'title',
    'text_no',
    'author',
    'story',
    'text_id',
    'word_count_nltk'
]
````

````python
df_short_stories_corpus = df_short_stories_sample_balanced[needed_columns].copy()
df_short_stories_corpus
````

## Create the column `file_id` and reorder the columns

````python
df_short_stories_corpus['file_id'] = 'f' + df_short_stories_corpus.index.astype(str).str.zfill(6)
````

````python
# Reorder the columns
first_columns = [
    'text_no',
    'text_id',
    'file_id'
]

df_short_stories_corpus = df_short_stories_corpus[first_columns + [col for col in df_short_stories_corpus.columns if col not in first_columns]]
````

````python
df_short_stories_corpus
````

## Export to a file

````python
df_short_stories_corpus.to_json(f"{output_directory}/{dataset_filename_2}.jsonl", orient='records', lines=True)
````

````python
df_short_stories_corpus.to_excel(f"{output_directory}/{dataset_filename_2}.xlsx", index=False)
````

## Extract the `story` column to individual files

### Import the data into a DataFrame

````python
df_short_stories_corpus = pd.read_json(f"{output_directory}/{dataset_filename_2}.jsonl", lines=True)
````

### Extract the `story` column to individual files

````python
for index, row in df_short_stories_corpus.iterrows():
    # Constructing the file name based on row data and index
    file_name = f"{output_directory}/{row['file_id']}.txt"

    # Writing the 'Short Story Text' content to the file
    with open(file_name, 'w', encoding='utf-8') as file:
        file.write(row['story'])
````
