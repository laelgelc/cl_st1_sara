The purpose of this document is to extract a sample of 1000 texts from the [4000 stories with sentiment analysis dataset](https://brunel.figshare.com/articles/dataset/4000_stories_with_sentiment_analysis_dataset/7712540?file=14357549) balanced across `author`.

## What is the `4000 stories with sentiment analysis dataset`?

This dataset presents 4,000 short stories that have been classified in terms of their emotional content and semantic structure. Emotional content was calculated using the valence, arousal and dominance norms in Warriner et al. (2014). Semantic structure was derived using the doc2vec algorithm, which classifies each text as a 300-place vector. The authors created this dataset as part of a study of the impact of narrative literature on mental health.

## `Contemporary authors` subcorpus

Parse contemporary short story metadata

```python
from pathlib import Path
import pandas as pd

cont_stories_dir = Path("corpus/00_sources/additional_contemporary_authors")

records = []

for file_path in sorted(cont_stories_dir.glob("*.txt")):
    parts = [part.strip() for part in file_path.stem.split(" - ")]

    if len(parts) == 3:
        title, collection_title, author = parts
    elif len(parts) == 2:
        title, author = parts
        collection_title = None
    else:
        title = parts[0]
        collection_title = None
        author = None

    records.append({
        "title": title,
        "collection_title": collection_title,
        "author": author,
    })

df_cont_short_stories = pd.DataFrame(records)

df_cont_short_stories
```

Add story IDs

```python
df_cont_short_stories["story_id"] = [
    f"cont_story_{i:04d}" for i in range(1, len(df_cont_short_stories) + 1)
]

df_cont_short_stories.head()
```

Export contemporary stories as individual text files

```python
from pathlib import Path
import shutil

source_dir = Path("corpus/00_sources/additional_contemporary_authors")
output_dir = Path("corpus/cont_stories")
output_dir.mkdir(parents=True, exist_ok=True)

for _, row in df_cont_short_stories.iterrows():
    if pd.notna(row["collection_title"]):
        source_filename = f"{row['title']} - {row['collection_title']} - {row['author']}.txt"
    else:
        source_filename = f"{row['title']} - {row['author']}.txt"

    source_path = source_dir / source_filename
    output_path = output_dir / f"{row['story_id']}.txt"

    shutil.copy2(source_path, output_path)

print(f"Stories copied: {len(list(output_dir.glob('*.txt')))}")
```

Normalize quotation marks in story files.

Parse every `.txt` file in `corpus/4000_stories/` and `corpus/cont_stories/`, and write a copy to `corpus/4000_stories_fixed/` and `corpus/cont_stories_fixed/`, replacing:
- `“` → `"`
- `”` → `"`
- `‘` → `'`
- `’` → `'`

Return a small summary (files scanned/written and total replacements).

```python
from __future__ import annotations

from pathlib import Path

SOURCE_TO_DESTINATION_DIRS = {
    Path("corpus") / "4000_stories": Path("corpus") / "4000_stories_fixed",
    Path("corpus") / "cont_stories": Path("corpus") / "cont_stories_fixed",
}

TRANSLATION_TABLE = str.maketrans({
    "“": '"',
    "”": '"',
    "‘": "'",
    "’": "'",
})

summary_quotes = {
    "directories_processed": 0,
    "files_scanned": 0,
    "source_txt_files": 0,
    "files_written": 0,
    "total_replacements": 0,
}

for src_dir, dst_dir in SOURCE_TO_DESTINATION_DIRS.items():
    if not src_dir.exists():
        raise FileNotFoundError(f"Source directory not found: {src_dir.resolve()}")

    dst_dir.mkdir(parents=True, exist_ok=True)
    summary_quotes["directories_processed"] += 1

    for src_path in sorted(src_dir.iterdir()):
        summary_quotes["files_scanned"] += 1

        if not (src_path.is_file() and src_path.suffix.lower() == ".txt"):
            continue

        summary_quotes["source_txt_files"] += 1

        raw = src_path.read_text(encoding="utf-8", errors="replace")
        fixed = raw.translate(TRANSLATION_TABLE)

        summary_quotes["total_replacements"] += (
                raw.count("“")
                + raw.count("”")
                + raw.count("‘")
                + raw.count("’")
        )

        dst_path = dst_dir / src_path.name
        dst_path.write_text(fixed, encoding="utf-8")

        summary_quotes["files_written"] += 1

summary_quotes
```

Add word counts from fixed story files

```python
from pathlib import Path
import re

fixed_dirs = {
    "df_4000_short_stories": Path("corpus/4000_stories_fixed"),
    "df_cont_short_stories": Path("corpus/cont_stories_fixed"),
}

word_pattern = re.compile(r"\b[\w'-]+\b")

for df_name, fixed_dir in fixed_dirs.items():
    if not fixed_dir.exists():
        raise FileNotFoundError(f"Fixed directory not found: {fixed_dir.resolve()}")

    word_counts = {}

    for file_path in sorted(fixed_dir.glob("*.txt")):
        story_id = file_path.stem
        text = file_path.read_text(encoding="utf-8", errors="replace")
        word_counts[story_id] = len(word_pattern.findall(text))

    if df_name == "df_4000_short_stories":
        df_4000_short_stories["word_count"] = df_4000_short_stories["story_id"].map(word_counts)
    elif df_name == "df_cont_short_stories":
        df_cont_short_stories["word_count"] = df_cont_short_stories["story_id"].map(word_counts)

print("4000 stories word counts added:", df_4000_short_stories["word_count"].notna().sum())
print("Contemporary stories word counts added:", df_cont_short_stories["word_count"].notna().sum())
```

### Flag stories within the word count IQR

```python
q1 = df_short_stories["word_count"].quantile(0.25)
q3 = df_short_stories["word_count"].quantile(0.75)

df_short_stories["within_iqr"] = df_short_stories["word_count"].between(q1, q3, inclusive="both")

print(f"Q1: {q1}")
print(f"Q3: {q3}")
print(df_short_stories["within_iqr"].value_counts())
```

List the authors and their respective number of stories that fall within the IQR

```python
authors_within_iqr = (
    df_short_stories[df_short_stories["within_iqr"]]
    .groupby("author")
    .size()
    .reset_index(name="story_count")
    .sort_values(["story_count", "author"], ascending=[False, True])
)

authors_within_iqr
```

### Select balanced IQR-filtered stories by author

This cell creates a boolean `selected` column in `df_short_stories`. Only stories already marked as `within_iqr` are eligible for selection. For authors with **13 or fewer** stories within the IQR, all eligible stories are selected. For authors with **more than 13** eligible stories, exactly **13 stories** are randomly sampled. Stories outside the IQR remain unselected.

```python
RANDOM_SEED = 42
MAX_STORIES_PER_AUTHOR = 13

df_short_stories["selected"] = False

eligible_stories = df_short_stories[df_short_stories["within_iqr"]]

selected_indices = (
    eligible_stories
    .groupby("author", group_keys=False)
    .apply(
        lambda group: group.sample(
            n=min(len(group), MAX_STORIES_PER_AUTHOR),
            random_state=RANDOM_SEED
        )
    )
    .index
)

df_short_stories.loc[selected_indices, "selected"] = True

selection_summary = (
    df_short_stories[df_short_stories["selected"]]
    .groupby("author")
    .size()
    .reset_index(name="selected_story_count")
    .sort_values(["selected_story_count", "author"], ascending=[False, True])
)

print(f"Total selected stories: {df_short_stories['selected'].sum()}")
selection_summary
```
