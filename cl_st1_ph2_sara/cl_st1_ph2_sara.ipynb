## Reorganise files into batches for tagging with Biber Tagger

### `corpus/05_tagged/03_llm_prep`

```python
import os
import shutil

# --- Configuration ---
# Configurable input directory relative to the current working directory
input_dir = "corpus/05_tagged/03_llm_prep"
batch_size = 250
# ---------------------

def reorganise_into_batches(directory, size):
    # Ensure the directory exists
    if not os.path.exists(directory):
        print(f"Error: Directory '{directory}' does not exist.")
        return

    # Get all files in the directory (excluding any existing directories)
    files = [f for f in os.listdir(directory) if os.path.isfile(os.path.join(directory, f))]

    # Sort files sequentially by filename
    files.sort()

    total_files = len(files)
    if total_files == 0:
        print(f"No files found in '{directory}'.")
        return

    print(f"Found {total_files} files in '{directory}'.")

    # Calculate the total number of batches
    total_batches = (total_files + size - 1) // size

    # Move files into batch subdirectories
    for i in range(0, total_files, size):
        batch_files = files[i:i + size]
        batch_index = (i // size) + 1
        batch_dir_name = f"batch_{batch_index}"
        batch_dir_path = os.path.join(directory, batch_dir_name)

        # Create the batch directory
        os.makedirs(batch_dir_path, exist_ok=True)

        # Move the files into the new directory
        for f in batch_files:
            src = os.path.join(directory, f)
            dst = os.path.join(batch_dir_path, f)
            shutil.move(src, dst)

        print(f"Created '{batch_dir_name}' and moved {len(batch_files)} files into it.")

    print(f"Successfully reorganised {total_files} files into {total_batches} batches.")

# Run the reorganisation
reorganise_into_batches(input_dir, batch_size)
```

## Merge LLM Tag Counts from Batches

### `corpus/05_tagged/03_llm_prep`

```python
from pathlib import Path

# Configurable paths
base_dir = Path("corpus/05_tagged/03_llm_prep")

input_files = [
    base_dir / "batch_1/tagcount/03_llm_b1_counts.txt",
    base_dir / "batch_2/tagcount/03_llm_b2_counts.txt",
    base_dir / "batch_3/tagcount/03_llm_b3_counts.txt",
    base_dir / "batch_4/tagcount/03_llm_b4_counts.txt",
    base_dir / "batch_5/tagcount/03_llm_b5_counts.txt"
]

output_file = base_dir / "tagcount/03_llm_counts.txt"

# Create the output directory if it doesn't exist
output_file.parent.mkdir(parents=True, exist_ok=True)

# Concatenate files
with output_file.open("w", encoding="utf-8") as outfile:
    for file_path in input_files:
        if file_path.exists():
            content = file_path.read_text(encoding="utf-8")
            outfile.write(content)

            # Ensure a trailing newline is present between concatenations
            # to prevent merging the last line of one file with the first of the next
            if not content.endswith("\n"):
                outfile.write("\n")
        else:
            print(f"Warning: Input file not found: {file_path}")

print(f"Successfully concatenated files into: {output_file}")
```
