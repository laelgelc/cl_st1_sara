## Export SAS CSV Results as Markdown and LaTeX Tables

```python
from pathlib import Path

import pandas as pd


# Directory where the rendered tables will be stored
sas_results_dir = Path("sas_results")
sas_results_dir.mkdir(parents=True, exist_ok=True)


def title_to_markdown_heading(title: str) -> str:
    return f"# {title}\n\n"


def title_to_latex_caption(title: str) -> str:
    escaped = (
        title.replace("\\", r"\textbackslash{}")
        .replace("&", r"\&")
        .replace("%", r"\%")
        .replace("$", r"\$")
        .replace("#", r"\#")
        .replace("_", r"\_")
        .replace("{", r"\{")
        .replace("}", r"\}")
    )
    return escaped


written_tables = []

for title, csv_path in sas_outputs.items():
    csv_path = Path(csv_path)

    if not csv_path.exists():
        print(f"Skipping missing file: {csv_path}")
        continue

    if csv_path.suffix.lower() != ".csv":
        print(f"Skipping non-CSV file: {csv_path}")
        continue

    df = pd.read_csv(csv_path)

    output_stem = csv_path.stem
    markdown_path = sas_results_dir / f"{output_stem}.md"
    latex_path = sas_results_dir / f"{output_stem}.tex"

    markdown_table = df.to_markdown(index=False)
    markdown_path.write_text(
        title_to_markdown_heading(title) + markdown_table + "\n",
        encoding="utf-8",
    )

    latex_table = df.to_latex(
        index=False,
        longtable=True,
        caption=title_to_latex_caption(title),
        label=f"tab:{output_stem.replace('_', '-')}",
        escape=True,
    )
    latex_path.write_text(latex_table, encoding="utf-8")

    written_tables.append(
        {
            "title": title,
            "source_csv": str(csv_path),
            "markdown": str(markdown_path),
            "latex": str(latex_path),
        }
    )

written_tables_df = pd.DataFrame(written_tables)
written_tables_df
```
