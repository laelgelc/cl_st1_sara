## Prepare the corpus for tagging with Biber Tagger

### Detect and replace `curly inverted commas` by `straight quotes` in the corpus files
- Parse each text file in the `corpus/01_CEP` directory and copy it to the `corpus/02_CEP_fixed` replacing `curly inverted commas` by `straight quotes` as follows:
    - `“` by `"`
    - `”` by `"`
    - `‘` by `'`
    - `’` by `'`

#### Task: Fix curly quotes in corpus files (02 → 03)
Parse every `.txt` file in `corpus/01_CEP/` and write a copy to `corpus/02_CEP_fixed/`, replacing:
- `“` → `"`
- `”` → `"`
- `‘` → `'`
- `’` → `'`

Return a small summary (files scanned/written and total replacements).

```python
from __future__ import annotations

from pathlib import Path

SRC_DIR_QUOTES = Path("corpus") / "01_CEP"
DST_DIR_QUOTES = Path("corpus") / "02_CEP_fixed"

if not SRC_DIR_QUOTES.exists():
    raise FileNotFoundError(f"Source directory not found: {SRC_DIR_QUOTES.resolve()}")

DST_DIR_QUOTES.mkdir(parents=True, exist_ok=True)

TRANSLATION_TABLE = str.maketrans({
    "“": '"',
    "”": '"',
    "‘": "'",
    "’": "'",
})

summary_quotes = {
    "files_scanned": 0,
    "source_txt_files": 0,
    "files_written": 0,
    "total_replacements": 0,
}

for src_path in sorted(SRC_DIR_QUOTES.iterdir()):
    summary_quotes["files_scanned"] += 1
    if not (src_path.is_file() and src_path.suffix.lower() == ".txt"):
        continue

    summary_quotes["source_txt_files"] += 1
    raw = src_path.read_text(encoding="utf-8", errors="replace")

    fixed = raw.translate(TRANSLATION_TABLE)

    # count replacements (sum of per-character deltas)
    summary_quotes["total_replacements"] += (
            raw.count("“") + raw.count("”") + raw.count("‘") + raw.count("’")
    )

    (DST_DIR_QUOTES / src_path.name).write_text(fixed, encoding="utf-8")
    summary_quotes["files_written"] += 1

summary_quotes
```

### Insert a space between digit and letter and wrap every 10 words within each text line

```python
import os
import re

# Define input and output directories
input_directory = 'corpus/02_CEP_fixed'
output_directory = 'corpus/03_CEP_fixed'

# Create the output directory if it doesn't exist
os.makedirs(output_directory, exist_ok=True)

# Iterate through all files in the input directory
for filename in os.listdir(input_directory):
    if filename.endswith(".txt"):
        input_path = os.path.join(input_directory, filename)
        output_path = os.path.join(output_directory, filename)

        try:
            with open(input_path, 'r', encoding='utf-8') as f:
                # Read all lines to preserve original structure
                original_lines = f.readlines()

            processed_output = []

            for line in original_lines:
                # 1. Insert space between digit and letter (e.g., "1a" -> "1 a")
                fixed_line = re.sub(r'(\d)([a-zA-Z])', r'\1 \2', line)

                # Strip whitespace to process words, but check if line was just empty space
                stripped_line = fixed_line.strip()

                if not stripped_line:
                    # If the original line was empty (e.g. a paragraph break), preserve it
                    processed_output.append("")
                    continue

                # 2. Wrap every 10 words WITHIN the current line
                words = stripped_line.split()

                # Chunk the words into groups of 10
                for i in range(0, len(words), 10):
                    chunk = words[i:i+10]
                    processed_output.append(" ".join(chunk))

            # Join all processed lines with newlines
            final_output = "\n".join(processed_output)

            # Save to the output file
            with open(output_path, 'w', encoding='utf-8') as f:
                f.write(final_output)

            print(f"Processed: {filename}")

        except Exception as e:
            print(f"Error processing {filename}: {e}")

print("All files processed.")
```
